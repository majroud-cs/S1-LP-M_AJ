# TP1 — Partie ExedBike

```python
import numpy as np
import matplotlib.pyplot as plt
from autograd import grad
import autograd.numpy as anp
import pulp

np.random.seed(42)
plt.rcParams["figure.figsize"] = (6, 4)
```

## Partie ExedBike (TP1) — Question 1

Reprise du code de descente de gradient et fit de la courbe selon une régression log-log


```python
def gradient_descent(cost_fun, w_init, alpha, epochs, eps=1e-12):
    '''Minimise cost_fun par descente de gradient.

    ENTREES
    cost_fun : callable, fonction de coût J(w)
    w_init   : array-like, point initial
    alpha    : float, pas (learning rate)
    epochs   : int, nombre maximal d'itérations
    eps      : float, arrêt si la norme du gradient passe sous eps

    SORTIES
    w       : point final (np.array)
    history : liste des itérés successifs (y compris w_init)
    '''
    w = np.array(w_init, dtype=float)
    history = [w.copy()]
    g = grad(cost_fun)   # g(w) renvoie le gradient de cost_fun en w (autograd)

    for k in range(epochs):
        # 1) calculez gk = g(w) (convertissez en np.array) :
        gk = np.array(g(w), dtype=float)   # gradient en k
        # 2) si la norme de gk est < eps : "break" :
        if np.linalg.norm(gk) < eps:       # gradient < eps
            break
        # 3) sinon : mettez à jour w = w - alpha * gk :
        w = w - alpha * gk                 # pas de descente
        # 4) ajoutez une copie de w à "history" :
        history.append(w.copy())

    return w, history
```

```python
# Jeu de données synthétique (ne changez pas la graine : les résultats doivent être reproductibles)
rng = np.random.default_rng(7)
q = np.linspace(50, 2000, 40)
k_true, theta_true = 9500.0, 0.55
c_obs = k_true * q ** (-theta_true) * np.exp(rng.normal(0, 0.04, size=q.shape))

plt.figure()
plt.scatter(q, c_obs, s=15, label="coût unitaire observé")
plt.xlabel("volume cumulé (unités)")
plt.ylabel("coût unitaire (EUR)")
plt.title("ExedBike Urban — données de courbe d'apprentissage")
plt.legend()
plt.show()
```

```python
x = np.log(q)
y = np.log(c_obs)

# standardisation de log q : moyenne nulle, écart-type 1
x_mean, x_std = x.mean(), x.std()
x_z = (x - x_mean) / x_std


# J_fit
def J_fit(params):
    # erreur quadratique moyenne entre la droite p0 - p1 * x_z et y
    p0, p1 = params
    return anp.mean((p0 - p1 * x_z - y) ** 2)


# gradient_descent
w_final, hist = gradient_descent(J_fit, w_init=[0.0, 0.0], alpha=0.1, epochs=5000)
p0_hat, p1_hat = w_final

# dé-strandardisation : log c = (p0 + p1*x_mean/x_std) - (p1/x_std) * log q
theta_hat = p1_hat / x_std
k_hat = np.exp(p0_hat + p1_hat * x_mean / x_std)

print(f"{len(hist)-1} itérations")
print(f"k = {k_hat:.1f}, theta = {theta_hat:.3f}")

# nuage de points et courbe fittée
qq = np.linspace(q.min(), q.max(), 200)
plt.figure()
plt.scatter(q, c_obs, s=15, label="coût unitaire observé")
plt.plot(qq, k_hat * qq ** (-theta_hat), color="C1", label="fit")
plt.xlabel("volume cumulé (unités)")
plt.ylabel("coût unitaire (EUR)")
plt.title("ExedBike Urban — fit de la courbe d'apprentissage")
plt.legend()
plt.show()

# fonction de perte :
losses = [J_fit(w) for w in hist]
plt.figure()
plt.semilogy(losses)
plt.xlabel("itération")
plt.ylabel("EQM (log-log, standardisé)")
plt.title("Courbe de convergence")
plt.show()
```

### Question 1 — Pourquoi standardiser $\log q$ ?


La descente de gradient avance dans la direction de la pente la plus forte. Dans le cas où les lignes de niveau de l'erreur forment une vallée étroite et allongée, la pente ne pointe pas vers le fond de la vallée mais vers les flancs. Cela conduit à une descente en zigzague et une convergence lente. C'est le même problème qu'à l'Exercice 1.2, où la Hessienne avait des courbures très différentes selon les axes : le problème est alors qualifié de "mal conditionné".

Lorsqu'on regarde les données, $\log q$ va de 3,9 à 7,6 : elles ne passent donc pas par 0. L'ordonnée à l'origine $p_0$ correspond à $\log q = 0$, c'est-à-dire $q = 1$, qui n'est pas dans les données. Par conséquent, $p_0$ et la pente $p_1$ sont fortement liées : changer l'une oblige à changer l'autre pour rester sur le nuage de points, ce qui allonge la vallée.

Centrer $\log q$ (moyenne nulle) et le diviser par son écart-type place l'origine au milieu des données. $p_0$ devient alors le niveau moyen de la courbe, et $p_0$ et $p_1$ deviennent indépendants : les lignes de niveau sont des cercles, et la descente va droit vers le minimum.


En représentant l'historique de la descente de gradient dans les deux cas : standardisation et non standardisation, on peut voir que la descente de gradient est beaucoup plus aisée dans le cas standardisé.


#### Sans standardisation
La vallée est étroite et allongée. La descente peut donc passer d'un côté à l'autre et converger lentement vers le fond.

#### Avec standardisation
les lignes de niveaux sont beaucoup plus circulaires et concentriques. La pente pointe vers le fond de la vallée, chaque pas rapproche directement de la solution.

La solution est la même mais la convergence a été plus rapide et mieux assurée dans le cas standardisé.

```python
# fonction de tracé tirée du TP : lignes de niveau et trajectoire de la descente
def show_history(history, cost_fun, xlim, ylim, title=""):
    a_values, b_values = np.meshgrid(
        np.linspace(*xlim), np.linspace(*ylim)
    )
    Z = np.array(
        [cost_fun([a, b]) for a, b in zip(a_values.flat, b_values.flat)]
    ).reshape(a_values.shape)
    plt.figure()
    plt.contour(a_values, b_values, Z, levels=20, cmap="viridis")
    h = np.array(history)
    plt.plot(h[:, 0], h[:, 1], "o-", color="crimson", markersize=3)
    plt.plot(h[0, 0], h[0, 1], "ks", label="départ")
    plt.plot(h[-1, 0], h[-1, 1], "k*", markersize=12, label="arrivée")
    plt.legend()
    plt.title(title)
    plt.xlabel("$p_0$ (ordonnée à l'origine)")
    plt.ylabel("$p_1$ (pente)")
    plt.show()


# sans standardisation : la perte sur log q brut
def J_brut(p):
    return anp.mean((p[0] - p[1] * x - y) ** 2)


# même pas (0,01) et même nombre de pas pour les deux cas
hist_brut = gradient_descent(J_brut, [0.0, 0.0], 0.01, 3000)[1]
hist_std = gradient_descent(J_fit, [0.0, 0.0], 0.01, 3000)[1]

show_history(hist_brut, J_brut, xlim=(-1, 11), ylim=(-0.8, 1.0),
             title="Sans standardisation")
show_history(hist_std, J_fit, xlim=(w_final[0] - 4, w_final[0] + 4),
             ylim=(w_final[1] - 4, w_final[1] + 4),
             title="Avec standardisation")
```

## Partie ExedBike (TP1) — Question 2

#### Résolution dans le cas général (continu)

```python
import pulp

prob = pulp.LpProblem("ExedBike_mix", pulp.LpMaximize)

# 1) variables de décision (Urban, Cargo, Sport), bornées >= 0
u = pulp.LpVariable("Urban", lowBound=0)
cg = pulp.LpVariable("Cargo", lowBound=0)
s = pulp.LpVariable("Sport", lowBound=0)

# 2) objectif : marge totale
prob += 120 * u + 200 * cg + 175 * s, "marge_totale"

# 3) contraintes de ressources (table du TP), chacune nommée
prob += 4 * u + 6 * cg + 5 * s <= 520, "main_oeuvre"
prob += 6 * u + 15 * cg + 9 * s <= 1100, "aluminium"
prob += 2 * u + 2 * cg + 3 * s <= 260, "machine"

# 4) résolution
prob.solve(pulp.PULP_CBC_CMD(msg=False))

# 5) affichage : mix, marge, contraintes saturées, prix duaux
print("Statut PuLP :", pulp.LpStatus[prob.status])
print("Urban =", u.value(), " Cargo =", cg.value(), " Sport =", s.value(), " marge =", pulp.value(prob.objective), "€")
for name, con in prob.constraints.items():
    print(f"  {name:14s} marge de manoeuvre = {con.slack:.2f}  (saturée : {abs(con.slack) < 1e-6})  prix dual = {con.pi + 0.0:.2f}")
```

#### Recherche de solutions entières :

```python
# Variante en nombres entiers : on ne peut produire que des unités entières.
# Utilisation de cat="Integer". PuLP passe en résolution entière mais toujours en CBC : séparation et évaluation (branch and bound, avec coupes).
prob_entier = pulp.LpProblem("ExedBike_mix_entier", pulp.LpMaximize)

u = pulp.LpVariable("Urban", lowBound=0, cat="Integer")
cg = pulp.LpVariable("Cargo", lowBound=0, cat="Integer")
s = pulp.LpVariable("Sport", lowBound=0, cat="Integer")

prob_entier += 120 * u + 200 * cg + 175 * s, "marge_totale"

prob_entier += 4 * u + 6 * cg + 5 * s <= 520, "main_oeuvre"
prob_entier += 6 * u + 15 * cg + 9 * s <= 1100, "aluminium"
prob_entier += 2 * u + 2 * cg + 3 * s <= 260, "machine"

prob_entier.solve(pulp.PULP_CBC_CMD(msg=False))

print("Statut PuLP :", pulp.LpStatus[prob_entier.status])
print("Urban =", u.value(), " Cargo =", cg.value(), " Sport =", s.value(), " marge =", pulp.value(prob_entier.objective), "€")
for name, con in prob_entier.constraints.items():
    print(f"  {name:14s} marge de manoeuvre = {con.slack:.2f}  (saturée : {abs(con.slack) < 1e-6})")
```

## Modèle de coût ajusté

En échelle log-log, le modèle $c(q) = k\,q^{-\theta}$ devient une **droite** : $\log c = \log k - \theta \log q$. Ajuster le modèle revient donc à réaliser une régression linéaire identifiant la droite qui fit au mieux le nuage de points. Sa pente vaut $-\theta$ et son ordonnée à l'origine vaut $\log k$.

**Paramètres ajustés** :

- $k \approx 9\,683$ € : le oût unitaire de la première unité produite ($q = 1$).
- $\theta \approx 0{,}555$ : Doubler le volume cumulé a pour effet de multiplié le coût unitaire par $2^{-\theta} \approx 0{,}68$, ce qui correspond à une baisse d'environ 32 %.

Finalement, le coût unitaire passe d'environ 1 103 € à $q = 50$ à environ 142 € à $q = 2\,000$. On retrouve dans ce modèle le phénomène d'économie d'échelle recherché en industrie.


```python
# régression dans le plan log-log : le modèle devient une droite
xx = np.linspace(x.min(), x.max(), 100)
plt.figure()
plt.scatter(x, y, s=15, label="log du coût observé")
plt.plot(xx, np.log(k_hat) - theta_hat * xx, color="C1", label="droite ajustée")
plt.xlabel("log q")
plt.ylabel("log c")
plt.title("Régression log-log : droite de pente −θ")
plt.legend()
plt.show()
```

## Prix duaux — lecture pour le directeur d'usine

Le prix dual d'une ressource est la marge gagnée pour une unité de cette ressource en plus, il n'est non nul que si la ressource est saturée.

On peut retrouver la signification de ce prix dual en simulant l'ajout d'une unité de ressource parmi les ressources saturées et on réalise la résolution de nouveau :

- **Main-d'œuvre : 25 € par heure.** Une heure de main-d'œuvre supplémentaire fait passer la marge de 17 850 € à 17 875 €. Une heure de plus est donc rentable tant que son coût reste inférieur à 25 €.
- **Temps machine : 0 € pour une heure, 70 € pour dix heures.** Il reste 4 h de réserve : une heure machine de plus ne modifie pas la solution. Dix heures de plus font passer la marge à 17 920 €. Il faut donc noter que le gain n'est pas linéaire, il arrive par paliers.
- **Aluminium : 0 €.** Il reste 17 kg inutilisés. Même 100 kg de ressources en plus ne changeront pas la solution trouvée. Ce n'est pas un levier pour augmenter la marge

Cette modélisation est très intéressante car elle conduit à mettre en évidence facilement quel est le levier de l'augmentation de marge, c'est ici l'augmentation de la main d'oeuvre dans un premier temps (embauche, interimaires, etc...) puis c'est le temps machine qui vient à manquer, ce sera donc le deuxième levier de marge une fois le premier augmenté. Quant à la quantité d'aluminum, celle-ci n'est pas un levier tant que le volume disponible des autres ressources n'a pas significativement augmenté par rapport à la situation présente.
